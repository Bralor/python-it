## Problémy?

---

Chyba, resp. upozornění na ni ještě **neznamená problém**.

Je to nejlepší nápověda, kterou dostaneš.

Čti ji **odspodu**. Poslední řádek říká *co* je špatně, řádek nad ním *kde*.

<br>

| Chyba | Česky | Typicky |
|:-|:-|:-|
| `SyntaxError` | Python kód vůbec nepřečetl | chybí závorka, uvozovka, dvojtečka |
| `NameError` | neznám takové jméno | překlep, nebo jsi proměnnou ještě nevytvořil |
| `TypeError` | špatný typ hodnoty | sčítáš text s číslem |
| `ValueError` | typ je dobrý, hodnota ne | `int("ahoj")` |
| `ZeroDivisionError` | dělení nulou | `10 / 0` |
| `IndexError` | index mimo rozsah | chceš 5. prvek ze tří |

## Přetypování

---

Vzpomínáš, že `input` vrací **text**? Zkus s ním počítat.

---

````python
# vek = input("Kolik ti je let? ")
# print(vek + 5)
````

`TypeError`. Python nevěděl, jestli `"15" + 5` má být `20` nebo `"155"`, a tak radši nedělá nic.

Je tedy potřeba, vyjasnit datové typy, aby se nedostal do neočekávaných situací.

<br>

| Funkce | Z čeho dělá co |
|:-:|:-|
| `int()` | text nebo desetinné číslo → **celé číslo** |
| `float()` | text nebo celé číslo → **desetinné číslo** |
| `str()` | cokoliv → **text** |

---

````python
vek = input("Kolik ti je let? ")
vek = int(vek)              # z textu uděláme číslo
budouci_vek = vek + 5

print("Za 5 let ti bude " + str(budouci_vek) + " let.")
````

````python
print(type(1))
````

Obě přetypování v jednom řádku se dají zapsat i takhle — kolem `input` prostě obalíš `int`:

---

````python
vek = int(input("Kolik ti je let? "))
print("Za 5 let ti bude " + str(vek + 5) + " let.")
````

### Zrada desetinných čísel

---

Tohle je jedno z nejslavnějších překvapení v programování:

---

````python
print(0.1 + 0.2)
````

Není to chyba Pythonu, chová se tak skoro každý jazyk.

Počítač počítá v **dvojkové soustavě** a některá desetinná čísla v ní nemají přesný zápis (podobně jako 1/3 nemá přesný desetinný zápis).

<br>

Pro nás z toho plyne jediné pravidlo: **výsledek na výpis zaokrouhli**. Slouží k tomu `round`:

---

````python
print(round(0.1 + 0.2, 1))   # na 1 desetinné místo
````

````python
print(round(22.221097470085356, 2))
````

Tím se dá dokončit BMI z minulého bloku:

---

````python
vyska = 1.75
hmotnost = 68

bmi = hmotnost / (vyska * vyska)

print("Tvoje BMI je " + str(round(bmi, 1)))
````

Na to má Python **knihovnu `decimal`**. Knihovna je balík hotového kódu, který si přitáhneš příkazem `import`.

````python
from decimal import Decimal

print(Decimal("0.1") + Decimal("0.2") + Decimal("0.3"))
````

> ⚠️ Všimni si uvozovek: `Decimal("0.1")`. Decimal na vstupu pracuje s textovou hodnotou, proto uvozovky.

---

````python
print(Decimal(0.1) + Decimal(0.2) + Decimal(0.3))   # bez uvozovek
````

<br>

S `Decimal` se počítá stejně jako s obyčejnými čísly:

---

````python
cena = Decimal("100.24")
sleva = Decimal("0.15")

print(cena * (1 - sleva))
````

A když potřebuješ zaokrouhlit **vždycky nahoru** (obchod to u ceny dělá rád), umí i to:

---

````python
from decimal import ROUND_CEILING

vysledna_cena = cena * (1 - sleva)

print(vysledna_cena.quantize(Decimal("1"), rounding=ROUND_CEILING))
````

Pro jednoduché příklady si vystačíš s `round`. Ale až budeš někdy počítat s penězi, vzpomeň si, že `float` na to není ideální.

### 💡 BONUS: co je proměnná doopravdy

---

Říkali jsme „proměnná je jméno pro hodnotu". Přesněji je to **odkaz** — cedule, která ukazuje na místo v paměti, kde hodnota leží.

Číslo toho místa ti prozradí zabudovaná funkce `id`:

---

````python
jmeno = "Matouš"

print(jmeno)
print(id(jmeno))   # číslo místa v paměti
````

Když proměnnou „změníš", hodnota v paměti se nepřepíše. Vznikne **nová** a cedule se přehodí jinam:

---

````python
jmeno = "Matouš"
print(id(jmeno))

jmeno = "Marek"
print(id(jmeno))   # jiné číslo = jiné místo
````

A naopak — dvě jména můžou ukazovat na **to samé** místo:

---

````python
a = "Matouš"
b = a

print(id(a))
print(id(b))   # stejné číslo
````

Konkrétní čísla budeš mít jiná než spolužák a po restartu jádra se změní. Na tom nezáleží. Důležité je, že **proměnná hodnotu neobsahuje, jen na ni ukazuje**.

#### Past: tuple dělá čárka, ne závorky

---

Tipni si, co vypíšou tyhle dvě buňky:

První závorka je jen **obyčejná závorka kolem textu**, jako v matematice. Teprve **čárka** dělá tuple.
